```python
clean_df = clean_df[clean_df["price"] >= 100000]
clean_df = clean_df[clean_df["distance"] <= 200000]
clean_df.loc[clean_df["mileage"] > 40, "mileage"] = np.nan
clean_df.loc[clean_df["engine_cc"] < 500, "engine_cc"] = np.nan
```

```python
REFERENCE_YEAR = 2026

clean_df["car_age"] = REFERENCE_YEAR - clean_df["year"]

# Extract brand from title
clean_df["brand"] = (
    clean_df["title"]
    .str.extract(r"^\d{4}\s+([A-Za-z]+)")[0]
    .str.strip()
)
```

```python
# Extract model name from title
# Format generally: YEAR BRAND MODEL ...

def extract_model(title):
    parts = str(title).split()

    # Remove year
    if parts and parts[0].isdigit():
        parts = parts[1:]

    # Brand
    if parts:
        brand = parts[0]
        parts = parts[1:]

    # Take first 1-2 words as model
    # Handles examples like:
    # Hyundai Venue
    # Tata Tiago
    # Honda City
    # Mercedes-Benz G-Class
    if len(parts) >= 2 and parts[1].lower() in [
        "class", "rover", "series"
    ]:
        model = parts[0] + " " + parts[1]
    elif parts:
        model = parts[0]
    else:
        model = "Unknown"

    return model


clean_df["model"] = clean_df["title"].apply(extract_model)

print("Brands:", clean_df["brand"].nunique())
print("Models:", clean_df["model"].nunique())

print("\nTop Brands:")
display(clean_df["brand"].value_counts().head(20))

print("\nTop Models:")
display(clean_df["model"].value_counts().head(20))
```

```python
# Better model extraction

def extract_brand_model(title):
    parts = str(title).split()

    # Remove year
    if parts and parts[0].isdigit():
        parts = parts[1:]

    # Known multi-word brands
    if len(parts) >= 2 and parts[0] == "Mercedes-Benz":
        brand = "Mercedes-Benz"
        model = parts[1]

    elif len(parts) >= 2 and parts[0] == "Land" and parts[1] == "Rover":
        brand = "Land Rover"
        model = parts[2] if len(parts) > 2 else "Unknown"

    elif len(parts) >= 2 and parts[0] == "Maruti" and parts[1] == "Suzuki":
        brand = "Maruti Suzuki"
        model = parts[2] if len(parts) > 2 else "Unknown"

    elif len(parts) >= 2 and parts[0] == "MG":
        brand = "MG"
        model = parts[1]

    else:
        brand = parts[0] if parts else "Unknown"
        model = parts[1] if len(parts) > 1 else "Unknown"

    return pd.Series([brand, model])


clean_df[["brand", "model"]] = clean_df["title"].apply(extract_brand_model)

print("Number of brands:", clean_df["brand"].nunique())
print("Number of models:", clean_df["model"].nunique())

print("\nTop brands:")
display(clean_df["brand"].value_counts().head(20))

print("\nTop models:")
display(clean_df["model"].value_counts().head(20))
```

```python
def fix_model(row):
    title = str(row["title"])

    # Specific multi-word models
    if "Grand i10" in title:
        return "Grand i10"

    if "Grand Vitara" in title:
        return "Grand Vitara"

    if "Innova Crysta" in title:
        return "Innova Crysta"

    if "Range Rover" in title:
        return "Range Rover"

    if "Range Rover Evoque" in title:
        return "Range Rover Evoque"

    if "Range Rover Velar" in title:
        return "Range Rover Velar"

    return row["model"]


clean_df["model"] = clean_df.apply(fix_model, axis=1)

print("Models after fixing:", clean_df["model"].nunique())

display(
    clean_df[
        ["title", "brand", "model", "year", "price"]
    ].sample(20, random_state=42)
)
```

```python
import numpy as np
import pandas as pd

# Car age
REFERENCE_YEAR = 2026
clean_df["car_age"] = REFERENCE_YEAR - clean_df["year"]

# Engine in litres
clean_df["engine_liters"] = clean_df["engine_cc"] / 1000

# Average distance driven per year
clean_df["km_per_year"] = (
    clean_df["distance"] / clean_df["car_age"].clip(lower=1)
)

# Automatic transmission flag
clean_df["is_automatic"] = (
    clean_df["transmission"]
    .astype(str)
    .str.lower()
    .str.contains("automatic")
    .astype(int)
)

# Electric vehicle flag
clean_df["is_ev"] = (
    clean_df["title"]
    .astype(str)
    .str.contains(r"\bEV\b|Electric", case=False, regex=True)
    .astype(int)
)

print("New features added successfully!")

display(
    clean_df[
        [
            "title",
            "year",
            "car_age",
            "distance",
            "km_per_year",
            "engine_cc",
            "engine_liters",
            "transmission",
            "is_automatic",
            "is_ev",
            "price"
        ]
    ].head(10)
)
```

```python
from sklearn.model_selection import train_test_split

# Final features
numeric_features = [
    "car_age",
    "distance",
    "engine_cc",
    "engine_liters",
    "mileage",
    "seats",
    "km_per_year",
    "owner_number",
    "is_automatic",
    "is_ev"
]

categorical_features = [
    "brand",
    "model",
    "transmission",
    "fuel"
]

features = numeric_features + categorical_features
target = "price"

X = clean_df[features].copy()
y = clean_df[target].copy()

# 80% training, 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Total rows :", len(X))
print("Training rows:", len(X_train))
print("Testing rows :", len(X_test))

print("\nNumber of features:", len(features))
print("\nFeatures:")
print(features)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Numeric preprocessing
numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

# Categorical preprocessing
categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        ))
    ]
)

# Combine both
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
    )

print("Preprocessing pipeline created successfully!")
```

```python
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Gradient Boosting model
gbr = GradientBoostingRegressor(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=3,
    min_samples_split=5,
    min_samples_leaf=2,
    random_state=42
)

# Complete pipeline
gbr_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", gbr)
    ]
)

# Train
gbr_pipeline.fit(X_train, y_train)

# Predict
y_pred = gbr_pipeline.predict(X_test)

# Evaluation
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("Gradient Boosting Results")
print("-" * 35)
print(f"MAE  : ₹{mae:,.2f}")
print(f"RMSE : ₹{rmse:,.2f}")
print(f"R²   : {r2:.4f}")
print(f"R² % : {r2 * 100:.2f}%")
```

```python
from sklearn.ensemble import ExtraTreesRegressor, RandomForestRegressor, HistGradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

models = {
    "Extra Trees": ExtraTreesRegressor(
        n_estimators=400,
        max_features=0.8,
        min_samples_leaf=1,
        random_state=42,
        n_jobs=-1
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=400,
        max_features=0.8,
        min_samples_leaf=1,
        random_state=42,
        n_jobs=-1
    ),

    "Hist Gradient Boosting": HistGradientBoostingRegressor(
        max_iter=300,
        learning_rate=0.05,
        max_leaf_nodes=31,
        l2_regularization=1.0,
        random_state=42
    )
}

results = []

for name, model in models.items():

    pipeline = Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            ("model", model)
        ]
    )

    pipeline.fit(X_train, y_train)

    predictions = pipeline.predict(X_test)

    mae = mean_absolute_error(y_test, predictions)
    rmse = np.sqrt(mean_squared_error(y_test, predictions))
    r2 = r2_score(y_test, predictions)

    results.append({
        "Model": name,
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2,
        "R2_percent": r2 * 100
    })

results_df = pd.DataFrame(results).sort_values(
    "R2",
    ascending=False
)

display(results_df)
```

```python
# Extract variant information from the title

def extract_variant(row):
    title = str(row["title"])
    year = str(row["year"])

    # Remove year from beginning
    title = title.replace(year, "", 1).strip()

    # Remove brand
    brand = str(row["brand"])
    title = title.replace(brand, "", 1).strip()

    # Remove model
    model = str(row["model"])
    title = title.replace(model, "", 1).strip()

    # Clean extra spaces
    variant = " ".join(title.split())

    return variant if variant else "Standard"


clean_df["variant"] = clean_df.apply(extract_variant, axis=1)

# Check examples
clean_df[["title", "brand", "model", "variant"]].head(20)
```

```python
# Fix multi-word models before extracting variant

def fix_model_name(row):
    title = str(row["title"])

    # Check longer names first
    if "Range Rover Evoque" in title:
        return "Range Rover Evoque"
    if "Range Rover Velar" in title:
        return "Range Rover Velar"
    if "Range Rover" in title:
        return "Range Rover"

    if "Grand i10" in title:
        return "Grand i10"
    if "Grand Vitara" in title:
        return "Grand Vitara"
    if "Innova Crysta" in title:
        return "Innova Crysta"
    if "Vitara Brezza" in title:
        return "Vitara Brezza"
    if "Urban cruiser" in title:
        return "Urban Cruiser"

    return row["model"]


clean_df["model"] = clean_df.apply(fix_model_name, axis=1)


# Re-extract variant using corrected model
def extract_variant(row):
    title = str(row["title"])
    year = str(row["year"])
    brand = str(row["brand"])
    model = str(row["model"])

    # Remove year
    title = title.replace(year, "", 1).strip()

    # Remove brand
    title = title.replace(brand, "", 1).strip()

    # Remove model
    title = title.replace(model, "", 1).strip()

    variant = " ".join(title.split())

    return variant if variant else "Standard"


clean_df["variant"] = clean_df.apply(extract_variant, axis=1)


# Check important examples
clean_df[
    clean_df["title"].str.contains(
        "Vitara Brezza|Urban cruiser|Grand i10|Innova Crysta|Range Rover",
        case=False,
        na=False
    )
][["title", "brand", "model", "variant"]].head(20)
```

```python
# Group rare variants to avoid too many one-hot categories

variant_counts = clean_df["variant"].value_counts()

# Variants appearing at least 3 times are kept
clean_df["variant_grouped"] = clean_df["variant"].where(
    clean_df["variant"].map(variant_counts) >= 3,
    "Other"
)

print("Unique variants before:", clean_df["variant"].nunique())
print("Unique variants after :", clean_df["variant_grouped"].nunique())

print("\nTop variants:")
display(clean_df["variant_grouped"].value_counts().head(20))
```
